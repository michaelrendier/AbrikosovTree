The Cayley-Dickson tower is the spine of the tree:

```
k=0  ℝ      σ=+1.00  dim=1    0 imaginary units   0 Fano planes  ← LEAVES (primes)
k=1  ℂ      σ=+0.75  dim=2    1 imaginary unit    0 Fano planes
k=2  ℍ      σ=+0.50  dim=4    3 imaginary units   0 Fano planes  ← gravastar / critical line
k=3  𝕆      σ=+0.25  dim=8    7 imaginary units   1 Fano plane
k=4  𝕊      σ= 0.00  dim=16  15 imaginary units   2 Fano planes  ← first ZD / equator
k=5  t_32   σ=−0.25  dim=32  31 imaginary units   4 Fano planes
k=6  t_64   σ=−0.50  dim=64  63 imaginary units   8 Fano planes
k=7  t_128  σ=−0.75  dim=128 127 imaginary units  16 Fano planes
k=8  T_256  σ=−1.00  dim=256 255 imaginary units  32 Fano planes ← ROOT
```

**Hurwitz 1898:** Only ℝ, ℂ, ℍ, 𝕆 are normed division algebras.  
At k=4 (𝕊, dim=16): zero-divisors first appear. `|ab|=|a||b|` fails.  
This is the algebraic source of FLT extinction (n≥3).  

**The angular quantum** at each level = 2π/dim.  
It halves with each CD doubling.  
At T_256: π/128 ≈ 1.41° — finest possible angular resolution before all roots merge to 1.

**n-sphere rotation (THE ANGLE, π/8 = 22.5°):**  
At each J_red level: rotate +22.5°.  
At each J_blue level: rotate −22.5°.  
Effect: prime paths straighten into radial spokes — geodesics on the sphere surface.  
The path that looks like a spiral in the base coordinates becomes a straight line  
when the local n-sphere is rotated to align the quadrant with the prime's N-shape.

## V(n): the n-ball volume transformer

````python
import matplotlib.pyplot as plt
import numpy as np
from fixed_point import v_nball

ns   = np.linspace(0, 256, 2000)
vols = [v_nball(float(n)) for n in ns]

# CD tower landmarks
landmarks = [(lv['k'], lv['dim'], lv['name'], lv['sigma']) for lv in tower]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Full range
axes[0].plot(ns, vols, color='#334466', linewidth=1.2)
for k, dim, name, sigma in landmarks:
    v = v_nball(float(dim))
    c = 'silver' if k == 0 else ('#8822bb' if k == 8 else ('#ffaa00' if k == 2 else ('#00cccc' if k == 4 else '#888888')))
    axes[0].axvline(dim, color=c, linestyle=':', linewidth=0.8)
    axes[0].scatter([dim], [v], color=c, zorder=5, s=40)
    if k in {0, 2, 4, 8}:
        axes[0].annotate(f'k={k}\n{name}', (dim, v), textcoords='offset points', xytext=(4, 4), fontsize=7)
axes[0].set_xlabel('n (dimension)')
axes[0].set_ylabel('V(n) — unit n-ball volume')
axes[0].set_title('V(n): The Transformer — from The Unit (V=1) to T_256 (V≈0)')
axes[0].set_xlim(0, 30)
axes[0].set_ylim(0, None)

# Detail: first 30 dimensions
ns_detail = np.linspace(0, 30, 300)
vols_detail = [v_nball(float(n)) for n in ns_detail]
axes[1].plot(ns_detail, vols_detail, color='#334466', linewidth=1.5)
for k, dim, name, sigma in landmarks:
    if dim <= 30:
        v = v_nball(float(dim))
        axes[1].scatter([dim], [v], s=60, zorder=5)
        axes[1].annotate(f'{name}\nσ={sigma:+.2f}', (dim, v),
                         textcoords='offset points', xytext=(5, -12), fontsize=8)
axes[1].axvline(5.257, color='red', linestyle='--', linewidth=1, label='n* (peak)')
axes[1].set_xlabel('n (dimension)')
axes[1].set_ylabel('V(n)')
axes[1].set_title('Detail: CD tower landmarks on V(n)')
axes[1].legend()

plt.tight_layout()
plt.savefig('02_vn_transformer.png', dpi=150)
plt.show()
````

````python
# Visualise 4 prime paths on the sphere (Space A top-down projection)
from mpl_toolkits.mplot3d import Axes3D

test_primes = [2, 11, 17, 47]
path_colors = ['gold', 'silver', '#cc4422', '#2244cc']

fig = plt.figure(figsize=(12, 5))

# Space A: sphere
ax1 = fig.add_subplot(121, projection='3d')
for p, col in zip(test_primes, path_colors):
    path = prime_tower_path(p)
    xs = [lv['sph']['x'] for lv in path['levels']]
    ys = [lv['sph']['y'] for lv in path['levels']]
    zs = [lv['sph']['z'] for lv in path['levels']]
    ax1.plot(xs, ys, zs, '-o', color=col, linewidth=1.5, markersize=4, label=f'p={p} e{p%16}')
ax1.set_title('Space A: Sphere paths (4 primes)')
ax1.legend(fontsize=7)

# Space B: plane stack
ax2 = fig.add_subplot(122)
for p, col in zip(test_primes, path_colors):
    path = prime_tower_path(p)
    xs = [lv['plane']['x'] for lv in path['levels']]
    zs = [lv['plane']['z'] for lv in path['levels']]
    ax2.plot(xs, zs, '-o', color=col, linewidth=1.5, markersize=5, label=f'p={p} e{p%16}')
for lv in tower:
    ax2.axhline(lv['sigma']*4, color='#aaaaaa', linewidth=0.5, linestyle=':')
    ax2.text(2.6, lv['sigma']*4, lv['name'], fontsize=7, va='center')
ax2.set_xlabel('x position (plane)')
ax2.set_ylabel('z = σ × 4')
ax2.set_title('Space B: Plane stack (x-z section, 4 primes)')
ax2.legend(fontsize=7)

plt.tight_layout()
plt.savefig('02_prime_paths.png', dpi=150)
plt.show()
````

## Angular quantum sequence and THE ANGLE

Each CD level halves the angular quantum.  
At k=4 (𝕊): angular quantum = π/8 = 22.5° = **THE ANGLE**.  
This is not a coincidence: THE ANGLE is the angular quantum of the first ZD level.  
Applying THE ANGLE (±22.5° rotation) to each level aligns all prime paths  
with the angular quantum of the sedenion — they become radial geodesics.

````python
print('Angular quantum sequence:')
print(f'  {"k":>3}  {"Name":<8}  {"dim":>5}  {"q_deg":>10}  Notes')
print('-' * 60)
for lv in tower:
    note = ''
    if lv['k'] == 4:  note = f' ← THE ANGLE ({lv["angular_q_deg"]:.1f}° = π/8)'
    if lv['k'] == 8:  note = ' ← finest resolution — T_256'
    if lv['k'] == 0:  note = ' ← degenerate (The Unit, no angle)'
    print(f'  k={lv["k"]}  {lv["name"]:<8}  {lv["dim"]:5d}  {lv["angular_q_deg"]:10.5f}°{note}')
print(f'\nTHE_ANGLE = {THE_ANGLE}°  (applied to each level to straighten prime paths)')
````
